```python
import numpy as np

# 1. The Data
vocab_size = 2
d_model = 3   # Tiny embedding dimension
d_ff = 4      # Tiny MLP hidden dimension
seq_len = 2   # We are processing 2 tokens

# Fake embeddings for "a" and "b" (Shape: 2x3)
X = np.array([[0.1, 0.5, -0.2],   # Token 'a'
              [-0.3, 0.2, 0.8]])  # Token 'b'

# The true next tokens: after 'a' comes 'b' (1), after 'b' comes 'a' (0)
target_indices = np.array([1, 0])

# Convert targets to one-hot vectors (Shape: 2x2)
# np.eye creates an identity matrix, and we index it with our targets!
y_one_hot = np.eye(vocab_size)[target_indices] 

# 2. The Weights (Initialized randomly)
W1 = np.random.randn(d_model, d_ff) * 0.1      # MLP Expand (3x4)
W_vocab = np.random.randn(d_ff, vocab_size) * 0.1 # Vocab Project (4x2)

learning_rate = 0.5
```

```python
for i in range(100):
    H_linear = X @ W1
    H_relu = np.maximum(H_linear, 0)

    # Logits
    z = H_relu @ W_vocab
    
    # Softmax
    z_sub_max = z - np.max(z, axis=-1, keepdims=True)
    z_sub_max_exp = np.exp(z_sub_max)
    z_sub_max_exp_sum = np.sum(z_sub_max_exp, axis=-1, keepdims=True)
    probs = z_sub_max_exp/z_sub_max_exp_sum

    loss = -np.mean(np.sum(y_one_hot * np.log(probs), axis=1))

    if i%10 == 0:
        print(f"---This is {i} run---")
        print(f"Loss is {loss}")


    # Loss

    dl_dz = probs - y_one_hot

    # Order is linear -> RELU -> Vocab Projection -> softmax -> loss

    # Vocab Projection - backpropogation
    dl_dwvocab = H_relu.T @ dl_dz
    dl_dhrelu = dl_dz @ W_vocab.T

    # RELU - backpropogation
    dl_dhlinear = dl_dhrelu * (H_linear > 0)
    
    dl_dw1 = X.T @ dl_dhlinear

    W1 = W1 - learning_rate * dl_dw1
    W_vocab = W_vocab - learning_rate * dl_dwvocab
```
