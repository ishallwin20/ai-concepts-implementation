```python
import numpy as np

seq_len = 3          # 3 tokens in our sequence
d_model = 8          # Embedding dimension
vocab_size = 5       # Tiny vocabulary: {0: "the", 1: "cat", 2: "sat", 3: "on", 4: "mat"}

# H represents the final hidden states of our 3 tokens after the Transformer Block
# Shape: (seq_len, d_model) -> (3, 8)
H = np.random.randn(seq_len, d_model)

# The Vocabulary Projection Matrix (Section 6.1)
# Shape: (d_model, vocab_size) -> (8, 5)
W_vocab = np.random.randn(d_model, vocab_size)

# The true next tokens for each of our 3 tokens (Section 6.4)
# e.g., after token 0 ("the"), the true next word is index 1 ("cat")
target_indices = np.array([1, 3, 0]) 
```

```python
# Logits
z = H @ W_vocab
print(f"Logits shape {z.shape}")

# Softmax Logits
z_max_sub = z - np.max(z, axis=-1, keepdims=True)
z_max_sub_exp = np.exp(z_max_sub)
z_max_sub_exp_sum = np.sum(z_max_sub_exp, axis=-1, keepdims=True)
probs = z_max_sub_exp/z_max_sub_exp_sum

# Verify Softmax 
# np.sum(probs, axis=-1)

# Numpy advance indexing
correct_probs = probs[np.arange(seq_len), target_indices]
log_correct_probs = np.log(correct_probs)
neg_log_correct_probs = -1 * log_correct_probs
loss = np.mean(neg_log_correct_probs)

print(loss)
```
