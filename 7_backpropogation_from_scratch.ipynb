```python
import numpy as np

# 1. The Input (shape: 1x2)
x = np.array([[2, 1]])

# 2. The Initial Weights (shape: 2x2)
W = np.array([[1, 3],
              [2, 4]])

# 3. The Target / Desired Output (shape: 1x2)
t = np.array([[5, 8]])

# 4. Learning Rate
mu = 0.1
```

```python
# y = x . W

y = x @ W
L = np.mean((t - y)**2)
print(f"Initial Loss is {L}")

dl_dy = y - t
print(f"gradient of L wrt y is {dl_dy}")

# Core is to find dl_dw
# xT . dl_dy = dl_dw

dl_dw = x.T @ dl_dy
print(f"gradient of L wrt W is {dl_dw}")

# Update W 
# W_new = W - mu . (dl_dw)

W_new = W - 0.1 * dl_dw
print(f"new W is {W_new}")

## Calculating new loss
y_new = x @ W_new
L_new = np.mean((t - y_new)**2)
print(f"New Loss is {L_new}")
```
